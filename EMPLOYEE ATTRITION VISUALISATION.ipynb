```python
df["Attrition"].value_counts().plot(kind="bar")

plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

```python
df.groupby(["Department", "Attrition"]).size()
```

```python
#Attrition rate in each dept
attrition_dept = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

attrition_dept
```

```python
ax = attrition_dept.plot(kind="bar")

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%")

plt.show()
```

```python
#attrition vs overtime
overtime_attrition = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

overtime_attrition
```

```python
overtime_yes = overtime_attrition["Yes"]
ax = overtime_yes.plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("OverTime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.ylim(0, 40)
ax.bar_label(ax.containers[0], fmt="%.1f%%")
plt.show()
```

```python
#attrition based on age
age_bins = [18, 25, 35, 45, 55, 65]
age_labels = ["18-25", "26-35", "36-45", "46-55", "56-65"]

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

age_attrition = pd.crosstab(
    df["AgeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

age_attrition
```

```python
age_yes = age_attrition["Yes"]

ax = age_yes.plot(kind="bar")

plt.title("Employee Attrition by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.ylim(0, 40)

ax.bar_label(ax.containers[0], fmt="%.1f%%")

plt.show()
```

```python
jobrole_attrition = pd.crosstab(
    df["JobRole"],
    df["Attrition"],
    normalize="index"
) * 100

jobrole_attrition
```

```python
#attrition with job role
jobrole_yes = jobrole_attrition["Yes"]

ax = jobrole_yes.plot(kind="bar")

plt.title("Employee Attrition by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 45)

ax.bar_label(ax.containers[0], fmt="%.1f%%")

plt.tight_layout()
plt.show()
```

```python
#attrition by income
income_bins = [0, 3000, 6000, 10000, 15000, 20000, 30000]
income_labels = [
    "Below 3K",
    "3K-6K",
    "6K-10K",
    "10K-15K",
    "15K-20K",
    "20K-30K"
]

df["IncomeGroup"] = pd.cut(
    df["MonthlyIncome"],
    bins=income_bins,
    labels=income_labels,
    include_lowest=True
)

income_attrition = pd.crosstab(
    df["IncomeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

income_attrition
```

```python
income_yes = income_attrition["Yes"]

ax = income_yes.plot(kind="bar")

plt.title("Employee Attrition by Monthly Income")
plt.xlabel("Monthly Income Group")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.ylim(0, 35)

ax.bar_label(ax.containers[0], fmt="%.1f%%")

plt.show()
```

```python
total_employees = df.shape[0]

print("Total Employees:", total_employees)
```

```python
employees_left = (df["Attrition"] == "Yes").sum()

print("Employees Who Left:", employees_left)
```

```python
attrition_rate = (employees_left / total_employees) * 100

print("Overall Attrition Rate:", round(attrition_rate, 2), "%")
```

```python
employees_stayed = (df["Attrition"] == "No").sum()

retention_rate = (employees_stayed / total_employees) * 100

print("Employees Stayed:", employees_stayed)
print("Retention Rate:", round(retention_rate, 2), "%")
```

```python
average_income = df["MonthlyIncome"].mean()

print("Average Monthly Income:", round(average_income, 2))
```

```python
average_age = df["Age"].mean()

print("Average Age:", round(average_age, 2))
```

```python
average_years = df["YearsAtCompany"].mean()

print("Average Years at Company:", round(average_years, 2))
```

```python
average_job_satisfaction = df["JobSatisfaction"].mean()

print(
    "Average Job Satisfaction:",
    round(average_job_satisfaction, 2)
)
```

```python
average_worklife = df["WorkLifeBalance"].mean()

print(
    "Average Work-Life Balance:",
    round(average_worklife, 2)
)
```

```python
overtime_employees = (df["OverTime"] == "Yes").sum()

overtime_rate = (overtime_employees / total_employees) * 100

print("Employees Working Overtime:", overtime_employees)
print("Overtime Rate:", round(overtime_rate, 2), "%")
```
